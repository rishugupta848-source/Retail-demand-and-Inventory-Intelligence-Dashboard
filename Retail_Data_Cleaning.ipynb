## 7.Data Cleaning

### *Data copy*

```python
customer_clean = customer.copy()
inventory_clean = inventory.copy()
promotions_clean = promotions.copy()
sales_clean = sales_transactions.copy()
sku_flags_clean = sku_flags.copy()
sku_clean = sku.copy()
store_clean = store.copy()
```

### *datetime conversion*

**Data Type Standardization**

Date fields were initially stored as string/object data types.

The relevant date columns were converted to Pandas datetime format using `pd.to_datetime()`.

Date columns included:

- Registration Date
- Last Restock Date
- Promotion Start Date
- Promotion End Date
- Sales Date
- Window Start
- Window End
- Store Opening Date

This conversion enables accurate date-based analysis and time-series visualization in subsequent stages.

```python
#customer
customer_clean["registration_date"] = pd.to_datetime(
    customer_clean["registration_date"],
    errors="coerce"
)
```

```python
#inventory
inventory_clean["last_restock_date"] = pd.to_datetime(
   inventory_clean["last_restock_date"],
    errors="coerce"
)
```

```python
#promotions
promotions_clean["start_date"] = pd.to_datetime(
   promotions_clean["start_date"],
    errors="coerce"
)
```

```python
#promotions
promotions_clean["end_date"] = pd.to_datetime(
   promotions_clean["end_date"],
    errors="coerce"
)
```

```python
#sales
sales_clean["date"] = pd.to_datetime(
   sales_clean["date"],
    errors="coerce"
)
```

```python
#sku flags
sku_flags_clean["window_start"] = pd.to_datetime(
  sku_flags_clean["window_start"],
    errors="coerce"
)
```

```python
#store
store_clean["opening_date"] = pd.to_datetime(
 store_clean["opening_date"],
    errors="coerce"
)
```

## 9.Relationship validation
### *Check foreign-key relationships*

The relationships between the datasets were validated using common identifier columns to ensure data consistency and integrity.

The key relationships checked were:

- `Sales.customer_id` → `Customer.cust_id`
- `Sales.sku_id` → `SKU.sku_id`
- `Sales.store_id` → `Store.store_id`
- `Sales.promo_id` → `Promotions.promo_id`
- `Inventory.sku_id` → `SKU.sku_id`
- `Inventory.store_id` → `Store.store_id`
- `SKU_Flags.sku_id` → `SKU.sku_id`

The identifier columns were checked for consistency, uniqueness where required, and matching values between related tables. These validations help ensure that the datasets can be reliably connected for further analysis and Power BI data modeling.



```python
#renaming customer_id-cust_id
sales_clean.rename(columns={'customer_id':'cust_id'},inplace=True)
```

```python
#sales-customer
sales_clean["cust_id"].isin(
    customer_clean["cust_id"]
).mean()
```

```python
#sales-SKU
sales_clean["sku_id"].isin(
    sku_clean["sku_id"]
).mean()
```

```python
#sales-store
sales_clean["cust_id"].isin(
    customer_clean["cust_id"]
).mean()
```

```python
#sales-promotions
sales_clean.loc[
    sales_clean["promo_id"].notna(),
    "promo_id"
].isin(promotions_clean["promo_id"]).mean()
```

**Thereby we got "1.0",meaning 100% valid relationships**
